```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

# 1. Load and clean your data (using the interactive input from earlier)
df = pd.read_csv("/kyr_querries.csv", sep=',', on_bad_lines='skip')
df = df.drop_duplicates(subset="query")

# 2. Split into train and test sets (e.g., 80% training, 20% testing)
train, test = train_test_split(df, test_size=0.2, random_state=42)

# 3. Vectorize the queries
vectorizer = TfidfVectorizer(
    lowercase=True,
    ngram_range=(1, 2)
)
X_train = vectorizer.fit_transform(train["query"])
X_test = vectorizer.transform(test["query"])

# 4. Train the model and make predictions
model = LogisticRegression(
    max_iter=1000
)
model.fit(
    X_train,
    train["intent"]
)
predictions = model.predict(X_test)


from sklearn.metrics import classification_report

print(
    classification_report(
        test["intent"],
        predictions
    )
)




def predict_intent(query):

    vector = vectorizer.transform([query])

    prediction = model.predict(vector)[0]

    return prediction

print(
    predict_intent(
        "My company has not given me my salary for 5 months"
    )
)

print(
    predict_intent(
        "My landlord is refusing to give my deposit back"
    )
)

print(
    predict_intent(
        "Someone made an unauthorized UPI transaction"
    )
)
```
